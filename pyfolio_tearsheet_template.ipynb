```python
# ✅ 1. 导入库和安装
# 如果尚未安装 Pyfolio，取消注释以下命令安装（推荐使用修复兼容性版本）
# pip install pyfolio-reloaded

import pyfolio as pf
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# ✅ 2. 构造示例收益率序列（日频）
# returns 必须是带日期索引的 pd.Series 类型

np.random.seed(42)
n_days = 504  # 两年交易日
returns = pd.Series(
    np.random.normal(loc=0.0003, scale=0.01, size=n_days),
    index=pd.date_range("2022-01-01", periods=n_days, freq="B")
)
returns.head()
```

```python
# ✅ 3. 快速生成完整策略诊断报告（含收益、风险、因子、分布图等）
pf.create_full_tear_sheet(returns)
```

```python
# ✅ 4. 获取绩效指标数据（以数值形式输出）
from pyfolio import timeseries
stats = timeseries.perf_stats(returns)
print(stats)
```

```python
# ✅ 5. 模拟交易记录 DataFrame（用于交易分析页）
transactions = pd.DataFrame([
    {"sid": "AAPL", "amount": 100, "price": 150.0, "dt": pd.Timestamp("2022-03-01")},
    {"sid": "AAPL", "amount": -100, "price": 155.0, "dt": pd.Timestamp("2022-03-10")},
]).set_index("dt")
transactions
```

```python
# ✅ 6. 生成交易分析报告（需配合 transactions）
pf.create_txn_tear_sheet(
    returns=returns,
    transactions=transactions
)
```

```python
# ✅ 7. 保存报告为 PDF 文件
from matplotlib.backends.backend_pdf import PdfPages

with PdfPages("tear_sheet_output.pdf") as pdf:
    pf.create_full_tear_sheet(returns)
    for i in plt.get_fignums():
        fig = plt.figure(i)
        pdf.savefig(fig)
    plt.close("all")
```
